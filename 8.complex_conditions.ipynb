Sometimes you can combine conditions with AND instead of nesting if statements

```python
#nested conditionals
amount=float(input("Enter the shopping amount:"))

if amount>=1000:
    membership=input("Are a premium member? (yes/no):")

    if membership=="yes":
        discount=0.20
    elif membership=="no":
        discount=0.10

else:
    discount=0.05

final_amount=amount-(amount*discount)
print("Discount:",discount*100,"%")
print("Final amount:",final_amount)
```

```python
amount=int(input("Enter the amount:"))

membership=input("Are you membership holder of the shop? yes/no")

if(amount>=1000 and membership=="yes"):
    discount=0.20
elif(amount>=1000 and membership=="no"):
    discount=0.10

else:
    discount=0.05

print("The discount percentage offered is:",discount*100)
print("The final amount to be paid by te customer is:",amount-(amount*discount))

```

If you need to remember the results of a condition check later in your code, use Boolean variables as flags


```python
assignment_submitted=bool(input("Enter the Boolean value:"))=="True"
attendance_sufficient=bool(input("Enter the Boolean value:"))=="True"

eligible=assignment_submitted and attendance_sufficient
if(eligible):
    print("Student can write the exam")
else:
    print("Student cannot write the exam")
```

```python
attendance=float(input("Enter the attendance percentage:"))
fees_paid=input("Have you paid the exam fees? (yes/no):")
print(attendance)
print(fees_paid)
eligible=attendance>=85 and fees_paid=="yes"

if eligible:
    print("The student can write the exam")
else:
    print("The student cannot write the exam")
```

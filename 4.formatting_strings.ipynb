```python
first_name="Christopher"
last_name="Harrison"

output="Hello, {}" " {}".format(first_name,last_name)
print(output)
```

```python
first_name="Christopher"
last_name="Harrison"

output="Hello, {1}" " {0}".format(first_name,last_name)
print(output)
#indexing the strings 0 the first string 1 the second string
# order wise
```

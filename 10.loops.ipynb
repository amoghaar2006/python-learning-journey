Loop through a collection

```python
for name in ["Susan", "David", "Bill", "Justin"]:
    print(name, end=" ")
```

```python
for i in range(0,2):
    print(i, end=" ")
```

```python
name=["Chistopoer","Susan"]
index=0
while index<len(name):
    print(name[index], end=" ")
    index+=1
```

```python
num=5
while num>=1:
    print(num, end=" ")
    num-=1
```

```python
people=["Christopher", "Susan"]

for name in people:
    print(name, end=" ")
```

```python
people=["Christopher", "Susan"]

index=0
while index<len(people):# Until the condition is true, the loop will continue to run
    print(people[index], end=" ")
    index+=1 
```

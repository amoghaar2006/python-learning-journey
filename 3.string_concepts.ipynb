You can combine strings with +

```python
first_name="Christopher"
last_name="Harrison"

full_name=first_name+ " " +last_name

print(full_name)

print("Hello" + " " + full_name)
```

You can use functions to modefiy strings

```python
sentence="The dog is named Sammy"
print(sentence.upper()) # converts all the char into uppercase
print(sentence.title()) # converts the first char of each word into uppercase
print(sentence.lower()) # converts all the char into lowercase
print(sentence.capitalize()) #converts the first char of the string as uppercase
print(sentence.count("a")) # counts the no of occurances of a substring in a string
```

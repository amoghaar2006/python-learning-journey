If an action depeends on a combination of conditions you can use nest if statements

```python
amount=float(input("Enter the shopping amount:"))

if amount>=1000:
    membership=input("Are a premium member? (yes/no):")

    if membership=="yes":
        discount=0.20
    elif membership=="no":
        discount=0.10

else:
    discount=0.05

final_amount=amount-(amount*discount)
print("Discount:",discount*100,"%")
print("Final amount:",final_amount)
```

```python
marks=int(input("Enter the marks obtained by the student:"))

if(marks>90):
    print("Grade of the student A")
elif(marks<=90 and marks>80):
    print("Grade of the student B")
elif(marks<=80 and marks>70):
    print("Grade of the student C")
else:
    print("Student is below average")
```

```python
units=int(input("Enter the number of units used by the household:"))
if(units>300):
    print("High electricity usage")
elif(units>200 and units<=300):
    print("Moderate electricity usage")
elif (units>100 and units<=200):
    print("Low electricity usage")
else:
    print("Very low electricity usage")
```

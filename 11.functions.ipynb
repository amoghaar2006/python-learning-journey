```python
def get_intial(name, force_uppercase=True):
    # Default parameter set for force_uppercase
# so while calling the function, if we do not pass any value for 
# force_uppercase, it will be set to True by defualt
    if force_uppercase:
        intial=name[0:1].upper()
    else:
        intial=name[0:1]

    return intial

first_name=input("Enter your first name:")
first_name_intial=get_intial(first_name,False)
print(first_name_intial)
```

```python
# in the same we can specify the parameter value by specifying the parameter while
# calling the function.
# in this case, the order doesn't matter, we can specify the parameter value in
# any order we want
first_name_intial=get_intial(force_uppercase=False, name=first_name)


# This method is useful when you have too many parameters in a function 
# where getting confused about the paramets values and their order is possible.
```

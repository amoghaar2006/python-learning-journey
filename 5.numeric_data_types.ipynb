You can do math with numbers

```python
first_num=6
second_num=2
print(first_num+second_num)
print(first_num*second_num)
print(first_num**second_num)
```

If you combine strings with numbers, Python gets confused

```python
days_in_feb=28
print(days_in_feb+"Days in February")

```

```python

#instead we can type casting
#converting the days_in_feb as a string then concatenating it
# with the second string
print(str(days_in_feb)+" " +"Days in February")
```

Numbers can be stored as strings
NUmbers stored as strings are treated as strings

```python
first_num=input("Enter a number")
second_num=input("Enter a number")

print(first_num+second_num)

print(int(first_num)+int(second_num))
# it is basically concatenating not adding the numbers
# because numbers are in the form of strings

# by default the input function gets the input in the form 
# of string data type
```

Instead in the input statement we can type cast the variable into int or float data type

```python
first=int(input("Enter a number"))
second=int(input("Enter a number"))

print(first)
print(second)
print("The sum of the two numbers:",first+second)

```

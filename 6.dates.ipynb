We often need current data and time when logging errors and saving

```python
# To get current date and time we need to use the datetime library
from datetime import datetime
current_date=datetime.now()

print("Today is:",str(current_date))
```

```python
# timedelta is used to define a period of time
from datetime import datetime,timedelta

today=datetime.now()
one_day=timedelta(days=1)
yesterday=today-one_day
print("Today is:",str(today))

print("Yesterday was:",str(yesterday))
```

Use date functions to control date formatting

```python
from datetime import datetime
current_date=datetime.now()

print(f"We are in {current_date.year} year")
print(f"We are in {current_date.month}th month of this year")
print(f"Todays date is:", str(current_date.day))
```

```python
birthday=input("When is your birthday (dd/mm/yyyy)")
current=datetime.strptime(birthday,"%d/%m/%Y")
print("Birthday:",birthday)
```

```python
today=datetime.now()


print("Today's Date is:",today.strftime("%d/%m/%Y"))
```

```python
today=datetime.now()

one_month=timedelta(days=30)

one_month_ago=today-one_month

print("The date of one month ago was:",one_month_ago.strftime("%d/%m/%Y"))
```

```python
today=datetime.now()

one_week=timedelta(weeks=1)

one_week_ago=today-one_week

print("The date of one week ago was:",one_week_ago.strftime("%d/%m/%Y"))
```

Lists are collections of items, they mutuable

```python
names=["Amoghaa R","Adarsha"]
scores=[]

scores.append(98) # Add a new itme to the end
scores.append(99)

print(names)
print(scores)
```

```python
print(scores[1])
```

Method	General syntax	Purpose
append()	scores.append(value)	    Adds a value to the end
insert()	scores.insert(index, value)	Inserts a value at a specific index
pop()	    scores.pop(index)	        Removes and returns the value at the index
pop()	    scores.pop()	            Removes and returns the last value
remove()	scores.remove(value)	    Removes the first occurrence of the specified value
